```python
Scenario 1:E-commerce Sales Analysis – Analysis of product categories, sales trends, and missing data insights.

Dataset: https://www.kaggle.com/datasets/carrie1/ecommerce-data
```

```python
print("\nK.Nivedha(24BAD082)")
print("\nEXPERIMENT 1")
print("\nSCENARIO 1: E-COMMERCE SALES DATA\n")

import pandas as pd
import matplotlib.pyplot as plt

#Load the dataset
df = pd.read_csv(r"C:\Users\nived\Downloads\archive (2)\data.csv", encoding="ISO-8859-1")

#first 5 rows
print("FIRST 5 ROWS:")
print(df.head())

#last 5 rows
print("\nLAST 5 ROWS:")
print(df.tail())

#Dataset information
print("\nDATASET INFO:")
df.info()

print("\nSTATISTICAL SUMMARY:")
print(df.describe())

print("\nMISSING VALUES IN EACH COLUMN:")
print(df.isnull().sum())

product_sales = df.groupby("Description")["Quantity"].sum().head(10)

#Bar chart
plt.figure()
product_sales.plot(kind='bar')
plt.xlabel("Product")
plt.ylabel("Quantity Sold")
plt.title("Top Selling Products")
plt.show()

#Line chart
plt.figure()
product_sales.plot(kind='line')
plt.xlabel("Product")
plt.ylabel("Quantity Sold")
plt.title("Sales Trend of Products")
plt.show()
```

```python
print("\nK.Nivedha(24BAD082)")
print("\nEXPERIMENT 1")
print("\nSCENARIO 2: HOSPITAL PATIENT RECORDS\n")

import pandas as pd
import matplotlib.pyplot as plt

#Load the dataset
df2 = pd.read_csv(r"C:\Users\nived\Downloads\diabetes.csv")

#first 5 rows
print("FIRST 5 ROWS:")
print(df2.head())

#Dataset information
print("\nDATASET INFO:")
df2.info()

#Check missing values
print("\nMISSING VALUES IN EACH COLUMN:")
print(df2.isnull().sum())

#Statistical summary
print("\nSTATISTICAL SUMMARY:")
print(df2.describe())

#Histogram 
plt.figure()
plt.hist(df2["Glucose"], bins=20)
plt.xlabel("Glucose Level")
plt.ylabel("Frequency")
plt.title("Glucose Level Distribution")
plt.show()

plt.figure()
plt.hist(df2["Age"], bins=10)
plt.xlabel("Age")
plt.ylabel("Count")
plt.title("Age Distribution")
plt.show()

#Boxplot 
plt.figure()
plt.boxplot(df2["Glucose"])
plt.title("Boxplot of Glucose Levels")
plt.show()

plt.figure()
plt.boxplot(df2["Age"])
plt.title("Boxplot of Age")
plt.show()

```

```python
Scenario 3: Housing Dataset Analysis – Investigation of housing attributes, correlations, and trends for price prediction.

Dataset: https://www.kaggle.com/datasets/harishkumardatalab/housing-price-prediction
```

```python
print("\nK.Nivedha(24BAD082)")
print("\nEXPERIMENT 1")
print("\nSCENARIO 3: HOUSING DATASET\n")

import pandas as pd
import matplotlib.pyplot as plt

#Load the dataset
df3 = pd.read_csv(r"C:\Users\nived\Downloads\Housing (1).csv")

#first 5 rows
print("FIRST 5 ROWS:")
print(df3.head())

#dataset information
print("\nDATASET INFO:")
df3.info()

#Check missing values
print("\nMISSING VALUES IN EACH COLUMN:")
print(df3.isnull().sum())

# Statistical summary
print("\nSTATISTICAL SUMMARY:")
print(df3.describe())

#Scatter plot
plt.figure()
plt.scatter(df3["area"], df3["price"])
plt.xlabel("Area")
plt.ylabel("Price")
plt.title("Area vs Price")
plt.show()

#Correlation matrix
numeric_df = df3.select_dtypes(include=['int64', 'float64'])

#Heatmap for correlation
corr = numeric_df.corr()

print("\nCORRELATION MATRIX:")
print(corr)


```

```python
Scenario 4: Customer Personality Analysis – Analysis of customer demographics, income levels, age distribution, and spending behavior to understand purchasing patterns.

Dataset: https://www.kaggle.com/datasets/imakash3011/customer-personality-analysis
```

```python
print("\nK.Nivedha(24BAD082)")
print("\nEXPERIMENT 1")
print("\nSCENARIO 4: CUSTOMER PERSONALITY\n")

import pandas as pd
import matplotlib.pyplot as plt

#Load dataset correctly
df = pd.read_csv(r"C:\Users\nived\Downloads\marketing_campaign.csv", sep="\t")

#Inspect dataset
print("FIRST 5 ROWS:")
print(df.head())

print("\nLAST 5 ROWS:")
print(df.tail())

print("\nDATASET INFORMATION:")
df.info()

print("\nSTATISTICAL DESCRIPTION:")
print(df.describe())

#Check missing values
print("\nMISSING VALUES:")
print(df.isnull().sum())

#Create Age column
df['Age'] = 2025 - df['Year_Birth']

#Bar plot 
plt.figure()
df['Age'].value_counts().head(10).plot(kind='bar')
plt.title("Age Distribution (Top 10)")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()

#Box plot
plt.figure()
df['Income'].plot(kind='box')
plt.title("Income Distribution")
plt.ylabel("Income")
plt.show()

#Calculate Total Spending
df['Total_Spending'] = (
    df['MntWines'] +
    df['MntFruits'] +
    df['MntMeatProducts'] +
    df['MntFishProducts'] +
    df['MntSweetProducts'] +
    df['MntGoldProds']
)

#Box plot 
plt.figure()
df['Total_Spending'].plot(kind='box')
plt.title("Customer Spending Pattern")
plt.ylabel("Total Spending")
plt.show()


```
